## 16.1 calculer la valeur de λ ( en utilisant que p(Ώ) = 1 )

soit un espace probabilité (Ώ , P) et X une variable aléatoir dont la loi est définie par ʎ ∈ R  ∀k = 1,2,3,4,5  P(X=k) = ʎ(3-k)²

```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
#...
def calculeʎ(a):
    return (1/10)*(3-a)**2
```

```python
def calculeʎliste(liste,ax=1):
    listeRes = np.zeros((2,len(liste)))
    listeRes[0] = liste
    for i in range(len(liste)):
        if calculeʎ(liste[i]) != 0. :
            listeRes[1][i] = calculeʎ(liste[i])
    return listeRes[ax]
```

```python
loi_x = np.zeros((2,4))
loi_x[0] = [1,2,4,5]
for i in range(len(loi_x[0])):
    loi_x[1][i] = (1/10)*(3-loi_x[0][i])**2
print(loi_x)
```

```python
loi_x_cum = np.zeros((2,5))
liste = [1,2,4,5]
val = calculeʎliste([1,2,4,5])
np.insert(loi_x_cum , 0 , [-1,0],axis=1)
#for i in range(len(val)):
#    np.insert(loi_x_cum, i ,[liste[i], val[i]], axis=1)
print(loi_x_cum)
```

```python
loi_x_cum = loi_x.copy()
loi_x_cum[1] = [np.sum(loi_x[1][:i+1]) for i in range(len(loi_x[1]))]
loi_x_cum = np.insert(loi_x_cum,0,[-1,0],axis=1)
loi_x_cum = np.insert(loi_x_cum,len(loi_x_cum[0]),[7,1],axis=1)
print(loi_x_cum)
```

```python
for i in range(len(loi_x_cum[0])-1):
    plt.plot([loi_x_cum[0][i],loi_x_cum[0][i+1]],[loi_x_cum[1][i],loi_x_cum[1][i]],"b")
plt.grid()
plt.show()
```
